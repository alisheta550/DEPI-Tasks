# Part 2

Compute the average and max price charged.  Store the results in variables called avg_price and max_price

```python
#Write your code here

# Calculate the average and maximum price
avg_price = df_data["price"].mean()
max_price = df_data["price"].max()

# View the results
print(f"Average Price: {avg_price:.4f}")
print(f"Maximum Price: {max_price:.2f}")
```

Compute the average price and average number of items purchased with only a single call to the mean() method. Store these averages in avg_price and avg_num_items.

```python
#Write you code here

# Compute both averages with a single call to mean()
averages = df_data[["price", "num_item"]].mean()

# Extract the individual average values
avg_price = averages["price"]
avg_num_items = averages["num_item"]

# View the results
print(f"Average Price: {avg_price:.4f}")
print(f"Average Num Items: {avg_num_items:.4f}")
```

Create a column called "total", which is the product of the price and num_item columns.

```python
#Write your code here

# Create the 'total' column by multiplying 'price' and 'num_item'
df_data["total"] = df_data["price"] * df_data["num_item"]

# Verify the new column
df_data[["price", "num_item", "total"]].head()
```

What is the avgerage price of the three most expensive purchases? 

```python
#Write your code here
# Option 1: Based on individual item unit price ('price')
top_3_prices = df_data.nlargest(3, "price")["price"].mean()
print(f"Average unit price: {top_3_prices:.4f}")

# Option 2: Based on total purchase cost ('total' = price * num_item)
top_3_totals = df_data.nlargest(3, "total")
avg_unit_price_of_top_totals = top_3_totals["price"].mean()
avg_total_cost_of_top_totals = top_3_totals["total"].mean()
```

Find the difference between the mean and median of the column total.  Use the describe() method and then index the dataframe it returns.

```python
#Write your code here

# Generate summary statistics for the 'total' column
total_stats = df_data["total"].describe()

# Index the Series to get mean and median (50%)
mean_total = total_stats["mean"]
median_total = total_stats["50%"]

# Calculate the difference
diff = mean_total - median_total

print(diff)
```

First, covert the column invoice time to a datetime column. Print out the new column types to make sure this was done correctly

```python
#Write your code here

# Convert the 'invoice_time' column to datetime
df_data["invoice_time"] = pd.to_datetime(
    df_data["invoice_time"], format="%m/%d/%y %H:%M"
)

# Print out the new column data types
print(df_data.dtypes)
```

Next, let's create columns called "DOW" and "Month", which store the corresponding weekday and month of the invoice respectively.  Print the head of the dataframe to make sure this was done correctly.

```python
#Write your code here

# Create DOW (Day of Week) and Month columns using datetime accessor dt
df_data["DOW"] = df_data["invoice_time"].dt.day_name()
df_data["Month"] = df_data["invoice_time"].dt.month_name()

# Display the first few rows to verify
df_data.head()
```

Create a variable called num_purchases_wed, which stores the number of purchases made on Wednesday?  You should use value_counts and then index the series it return.

```python
#Write your code here

# Get count of purchases per day of the week and index for 'Wednesday'
num_purchases_wed = df_data["DOW"].value_counts()["Wednesday"]

print(num_purchases_wed)
```

In this set of practice exercises, we will be working with a demographic data regarding the passengers aboard the Titanic.  Read in the data frame and use the head() method to check that it was read in correctly.

```python
#Write your code here

import pandas as pd

# Read in the Titanic dataset
df_titanic = pd.read_csv("Data/titanic.csv")

df_titanic.head()
```

Use the rename method to change the column "Name" to "Passenger_Name" and the column "Ticket" to "Ticket_Num".

```python
#Write your code here

df_titanic = df_titanic.rename(
    columns={"Name": "Passenger_Name", "Ticket": "Ticket_Num"}
)

# Alternatively, using inplace=True:

df_titanic.head()
```

Change the index of the df_titanic to the column PasengerID.

```python
#Write your code here

df_titanic = df_titanic.set_index("PassengerId")

# df_titanic.set_index('PassengerId', inplace=True)

df_titanic.head()
```

Select the name of passenger 896

```python
#Write your code here

passenger_896_name = df_titanic.loc[896, "Passenger_Name"]

print(passenger_896_name)
```

Change the index back to row numbers but making sure that we don't lose the PassengerId column. 

```python
#Write your code here

df_titanic = df_titanic.reset_index()

# df_titanic.reset_index(inplace=True)

df_titanic.head()
```

How many missing entries are there in the Age column? Use the isnull() method and recall than Booleans are exactly like 1s (True) and 0s (False).

```python
#Write you code here

missing_age_count = df_titanic["Age"].isnull().sum()

print(missing_age_count)
```

Compute the avg age of passengers ignoring the missing data.

```python
#Write your code here

avg_age = df_titanic["Age"].mean()

print(avg_age)
```

Using the fillna() method replace the missing values in the Age column with the mean.

```python
#Write your code here

# Calculate the mean age
mean_age = df_titanic["Age"].mean()

# Replace missing values in the 'Age' column with the mean
df_titanic["Age"] = df_titanic["Age"].fillna(mean_age)

# Verify that there are no remaining missing values
print(df_titanic["Age"].isnull().sum())
```

What is the average age of the 5 oldest passengers? The reset_index method will be helpful here.

```python
#Write your code here

top_5_oldest = df_titanic.sort_values(by="Age", ascending=False).reset_index(
    drop=True
)

avg_oldest_5 = top_5_oldest.head(5)["Age"].mean()

print(avg_oldest_5)
```
