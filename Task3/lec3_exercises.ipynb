````python
import math
````

````python
# 1
def area_of_circle(r):
    return math.pi * r * r

def perimeter_of_circle(r):
    return 2 * math.pi * r

print("Area of circle:", area_of_circle(10))
print("Perimeter of circle:", perimeter_of_circle(10))


# 2
def add_all_nums(*args):
    if all(isinstance(x, (int, float)) for x in args):
        return sum(args)
    return "All arguments must be numbers."

print("Sum of numbers:", add_all_nums(1, 2, 3, 4, 5))


# 3
def convert_celsius_2_fahrenheit(c):
    return (c * 9 / 5) + 32

print("Temperature in Fahrenheit:", convert_celsius_2_fahrenheit(25))


# 4
def check_season(month):
    month = month.lower()

    if month in ["september", "october", "november"]:
        return "Autumn"
    elif month in ["december", "january", "february"]:
        return "Winter"
    elif month in ["march", "april", "may"]:
        return "Spring"
    elif month in ["june", "july", "august"]:
        return "Summer"

print("Season:", check_season("January"))


# 5
def calculate_slope(x1, y1, x2, y2):
    return (y2 - y1) / (x2 - x1)

print("Slope:", calculate_slope(1, 2, 3, 6))


# 6
def solve_quadratic_eqn(a, b, c):
    d = b**2 - 4*a*c

    if d > 0:
        x1 = (-b + math.sqrt(d)) / (2*a)
        x2 = (-b - math.sqrt(d)) / (2*a)
        return x1, x2
    elif d == 0:
        return -b / (2*a)
    else:
        return "No real solutions"

print("Quadratic equation solutions:", solve_quadratic_eqn(1, -5, 6))


# 7
def print_list(items):
    print("List items:")
    for item in items:
        print(item)

print_list(["Python", "Java", "C++"])


# 9
def capitalize_list_items(items):
    return [item.capitalize() for item in items]

print("Capitalized list:", capitalize_list_items(["python", "java", "c++"]))


# 10
def add_item(items, item):
    items.append(item)
    return items

food_staff = ['Potato', 'Tomato', 'Mango', 'Milk']
print("Updated food list:", add_item(food_staff, 'Fungi'))

numbers = [2, 3, 7, 9]
print("Updated numbers list:", add_item(numbers, 5))


# 11
def remove_item(items, item):
    items.remove(item)
    return items

food_staff = ['Potato', 'Tomato', 'Mango', 'Milk']
print("Food list after removal:", remove_item(food_staff, 'Mango'))

numbers = [2, 3, 7, 9]
print("Numbers list after removal:", remove_item(numbers, 3))


# 12
def sum_of_numbers(n):
    return sum(range(1, n + 1))

print("Sum from 1 to 5:", sum_of_numbers(5))
print("Sum from 1 to 10:", sum_of_numbers(10))
print("Sum from 1 to 100:", sum_of_numbers(100))


# 13
def sum_of_odds(n):
    return sum(i for i in range(1, n + 1) if i % 2 != 0)

print("Sum of odd numbers:", sum_of_odds(10))


# 14
def sum_of_even(n):
    return sum(i for i in range(1, n + 1) if i % 2 == 0)

print("Sum of even numbers:", sum_of_even(10))
````

````python
# =========================
# Level 2
# =========================

# 1
def capitalize_list_items(items):
    return [item.capitalize() for item in items]

print("Capitalized list:", capitalize_list_items(["python", "java", "c++"]))


# 2
def add_item(items, item):
    items.append(item)
    return items

print("List after adding item:", add_item([1, 2, 3], 4))


# 3
def reverse_list(items):
    reversed_list = []

    for i in range(len(items) - 1, -1, -1):
        reversed_list.append(items[i])

    return reversed_list

print("Reversed list:", reverse_list([1, 2, 3, 4, 5]))
print("Reversed list:", reverse_list(["A", "B", "C"]))


# 4
def evens_and_odds(n):
    evens = 0
    odds = 0

    for i in range(n + 1):
        if i % 2 == 0:
            evens += 1
        else:
            odds += 1

    print("The number of odds are", odds)
    print("The number of evens are", evens)

evens_and_odds(100)


# 5
def calculate_mean(numbers):
    return sum(numbers) / len(numbers)


def calculate_median(numbers):
    numbers = sorted(numbers)
    n = len(numbers)

    if n % 2 == 0:
        return (numbers[n // 2 - 1] + numbers[n // 2]) / 2
    else:
        return numbers[n // 2]


def calculate_mode(numbers):
    frequency = {}

    for num in numbers:
        frequency[num] = frequency.get(num, 0) + 1

    max_frequency = max(frequency.values())

    modes = []

    for num, count in frequency.items():
        if count == max_frequency:
            modes.append(num)

    return modes


numbers = [1, 2, 2, 3, 4, 4, 4, 5]

print("Mean:", calculate_mean(numbers))
print("Median:", calculate_median(numbers))
print("Mode:", calculate_mode(numbers))


# 6
def print_list(items):
    print("List items:")

    for item in items:
        print(item)

print_list(["Python", "Java", "C++"])
````

````python
# =========================
# Level 3
# =========================
import keyword


# 1
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


print("Is 17 prime?", is_prime(17))
print("Is 20 prime?", is_prime(20))


# 2
def all_unique(items):
    return len(items) == len(set(items))


print("Are all items unique?", all_unique([1, 2, 3, 4]))
print("Are all items unique?", all_unique([1, 2, 2, 4]))


# 3
def same_data_type(items):
    if not items:
        return True

    first_type = type(items[0])

    for item in items:
        if type(item) != first_type:
            return False

    return True


print("Same data type:", same_data_type([1, 2, 3, 4]))
print("Same data type:", same_data_type([1, "2", 3]))


# 4

def is_valid_variable(variable):
    return variable.isidentifier() and not keyword.iskeyword(variable)


print("Is 'student_name' a valid variable?", is_valid_variable("student_name"))
print("Is '123student' a valid variable?", is_valid_variable("123student"))
print("Is 'class' a valid variable?", is_valid_variable("class"))
````
