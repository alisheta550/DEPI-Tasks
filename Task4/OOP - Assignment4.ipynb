### Exercise 10 — Implementing a Stack
Write a class `Stack` with an `items` list. Add `push(item)` (adds to the end) and `pop()` (removes and returns the last item, or `None` if empty). Push 1, 2, 3, then pop once and print both the popped value and the remaining stack.

```python
# write your code here
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if self.items:
            return self.items.pop()
        return None


stack = Stack()

stack.push(1)
stack.push(2)
stack.push(3)

popped = stack.pop()

print(popped)
print(stack.items)
```

### Exercise 12 — Implementing a Queue
Write a class `Queue` with an `items` list. Add `enqueue(item)` (adds to the end) and `dequeue()` (removes and returns the *first* item, or `None` if empty — note this is different from a Stack!). Enqueue "a", "b", "c", then dequeue once and print both the result and the remaining queue.

```python
# write your code here
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if self.items:
            return self.items.pop(0)
        return None


queue = Queue()

queue.enqueue("a")
queue.enqueue("b")
queue.enqueue("c")

dequeued = queue.dequeue()

print(dequeued)
print(queue.items)
```

### Exercise 13 — Objects containing other objects
Write a class `Movie` with `title` and `rating`. Write a class `MovieCollection` with a `movies` list, a method `add(movie)`, and a method `average_rating()` that returns the average rating of all movies (or 0 if empty). Add 3 movies with ratings 8, 6, 10 and print the average.

```python
# write your code here
class Movie:
    def __init__(self, title, rating):
        self.title = title
        self.rating = rating


class MovieCollection:
    def __init__(self):
        self.movies = []

    def add(self, movie):
        self.movies.append(movie)

    def average_rating(self):
        if not self.movies:
            return 0

        return sum(movie.rating for movie in self.movies) / len(self.movies)


collection = MovieCollection()

collection.add(Movie("Movie 1", 8))
collection.add(Movie("Movie 2", 6))
collection.add(Movie("Movie 3", 10))

print(collection.average_rating())
```

### Exercise 14 — Method chaining

```python
# write your code here
class Thermostat:
    def __init__(self):
        self.temp = 20

    def increase(self, amount):
        self.temp += amount
        return self

    def decrease(self, amount):
        self.temp -= amount
        return self


thermostat = Thermostat()

thermostat.increase(5).increase(3).decrease(2)

print(thermostat.temp)
```

### Exercise 15 — Inventory system with validation
Write a class `Inventory2` with a `stock` dictionary. `add_stock(item, qty)` should add to the existing quantity (or create it if new). `remove_stock(item, qty)` should only succeed (and return `True`) if there's enough stock, otherwise return `False` without changing anything. Add 50 then 20 more of "pen", try removing 30 (should succeed), then try removing 100 (should fail).

```python
# write your code here
class Inventory2:
    def __init__(self):
        self.stock = {}

    def add_stock(self, item, qty):
        self.stock[item] = self.stock.get(item, 0) + qty

    def remove_stock(self, item, qty):
        if self.stock.get(item, 0) >= qty:
            self.stock[item] -= qty
            return True
        return False


inventory = Inventory2()

inventory.add_stock("pen", 50)
inventory.add_stock("pen", 20)

print(inventory.remove_stock("pen", 30))
print(inventory.remove_stock("pen", 100))
```

### Exercise 16 — Class with calculated categories
Write a class `BMI` with `weight` and `height`. Add a method `calculate()` returning `weight / (height ** 2)`, and a method `category()` that returns `"Underweight"`, `"Normal"`, `"Overweight"`, or `"Obese"` based on the calculated value (use the same thresholds as the loops/functions lectures: 18.5, 25, 30). Test with weight=70, height=1.75.

```python
# write your code here
class BMI:
    def __init__(self, weight, height):
        self.weight = weight
        self.height = height

    def calculate(self):
        return self.weight / (self.height ** 2)

    def category(self):
        bmi = self.calculate()

        if bmi < 18.5:
            return "Underweight"
        elif bmi < 25:
            return "Normal"
        elif bmi < 30:
            return "Overweight"
        else:
            return "Obese"


bmi = BMI(70, 1.75)

print(bmi.calculate())
print(bmi.category())
```

### Exercise 18 — Tracking a running high score
Write a class `Game` with `score` and `high_score`, both starting at 0. `add_points(pts)` should increase `score`, and update `high_score` if `score` beats it. `reset()` should set `score` back to 0 **without** touching `high_score`. Add 50, then 30, then reset, then add 20 — print the final `score` and `high_score`.

```python
# write your code here
class Game:
    def __init__(self):
        self.score = 0
        self.high_score = 0

    def add_points(self, pts):
        self.score += pts

        if self.score > self.high_score:
            self.high_score = self.score

    def reset(self):
        self.score = 0


game = Game()

game.add_points(50)
game.add_points(30)
game.reset()
game.add_points(20)

print(game.score)
print(game.high_score)
```

### Exercise 19 — Payroll with overtime logic
Write a class `Employee3` with `name`, `hours`, and `rate`. Add a method `calculate_pay()`: normal pay is `hours * rate`, but any hours over 40 should be paid at 1.5× the rate (overtime). Test with an employee who worked 45 hours and one who worked 35 hours, both at rate 20 and 25 respectively.

```python
# write your code here
class Employee3:
    def __init__(self, name, hours, rate):
        self.name = name
        self.hours = hours
        self.rate = rate

    def calculate_pay(self):
        if self.hours > 40:
            normal_pay = 40 * self.rate
            overtime_pay = (self.hours - 40) * self.rate * 1.5
            return normal_pay + overtime_pay
        else:
            return self.hours * self.rate


employee1 = Employee3("Ali", 45, 20)
employee2 = Employee3("Ahmed", 35, 25)

print(employee1.calculate_pay())
print(employee2.calculate_pay())
```

### Exercise 20 — Comprehensive: a simple linked list
Write a class `Node2` with a `value` and a `next` (defaulting to `None`). Write a class `LinkedListSimple` with a `head` (starting as `None`) and a method `add(value)` that creates a new node and appends it to the end of the list (walking through `.next` links to find the end). Add a method `to_list()` that walks the list and returns a normal Python list of all the values, for easy printing. Add 10, 20, 30 and print the result of `to_list()`.

```python
# write your code here
class Node2:
    def __init__(self, value):
        self.value = value
        self.next = None


class LinkedListSimple:
    def __init__(self):
        self.head = None

    def add(self, value):
        new_node = Node2(value)

        if self.head is None:
            self.head = new_node
            return

        current = self.head

        while current.next is not None:
            current = current.next

        current.next = new_node

    def to_list(self):
        result = []
        current = self.head

        while current is not None:
            result.append(current.value)
            current = current.next

        return result


linked_list = LinkedListSimple()

linked_list.add(10)
linked_list.add(20)
linked_list.add(30)

print(linked_list.to_list())
```
